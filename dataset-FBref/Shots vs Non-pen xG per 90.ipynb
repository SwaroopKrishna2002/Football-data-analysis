```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = pd.read_csv("Midfielders Dataframe.csv")
```

```python
data.head()
```

```python
plt.figure(figsize=(15, 10))  # Adjust the figure size as per your preference
plt.scatter(data['Shots Total'], data['Non-Penalty xG'], alpha=0.7, edgecolors='k')
plt.xlabel('Shots Total (per 90 minutes)')
plt.ylabel('Non-Penalty xG (per 90 minutes)')
plt.title('Scatter Plot: Shots Total vs. Non-Penalty xG')
plt.grid(True)
plt.show()
```
