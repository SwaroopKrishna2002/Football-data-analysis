```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = pd.read_csv("Forwards Dataframe.csv")
```

```python
data.head()
```

```python
plt.figure(figsize=(15, 10))  # Adjust the figure size as per your preference
plt.scatter(data['Shots Total'], data['Shot-Creating Actions'], alpha=0.7, edgecolors='k')
plt.xlabel('Total Shots Taken (per 90 minutes)')
plt.ylabel('Shot Creating Actions (per 90 minutes)')
plt.title('Scatter Plot: Non-Penalty Goals vs. Non-Penalty xG')
plt.grid(True)
plt.show()
```
