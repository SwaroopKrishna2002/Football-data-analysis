```python
metric_team = ['Home_Team','Away_Team']
metric_score = ['Home_Team_Score','Away_Team_Score']
metric_possession = ['Home_Team_Possession','Away_Team_Possession']
metric_shots = ['Home_Team_Shots','Away_Team_Shots']
metric_shots_on_target = ['Home_Team_ShotsonTarget','Away_Team_ShotsonTarget']
metric_passes = ['Home_Team_Passes','Away_Team_Passes']
metric_fouls = ['Home_Team_Fouls','Away_Team_Fouls']
metric_yellow = ['Home_Team_Yellowcard','Away_Team_Yellowcard']
metric_red = ['Home_Team_Redcard','Away_Team_Redcard']
metric_offsides = ['Home_Team_Offside','Away_Team_Offside']
metric_corners = ['Home_Team_Corner','Away_Team_Corner']
```

```python
def extract_numeric_value(value):
    if isinstance(value, str):
        return float(value[:-1])
    return value
```

```python
while True:
    home_input = input("Enter a home team :").strip()
    away_input = input("Enter an away team :").strip()
    
    if home_input.lower()=='exit':
        break
        
    if away_input.lower()=='exit':
        break
        
    
    if home_input.lower() in home.str.lower().values and away_input.lower() in away.str.lower().values:
        match_data = data[(data['Home_Team'].astype(str).str.lower()==home_input.lower()) & (data['Away_Team'].astype(str).str.lower()==away_input.lower())]

        table_metrics = ["Score", "Possession (%)", "Shots", "Shots on Target", "Passes", "Fouls", "Yellow Cards", "Red Cards", "Offsides", "Corners"]
        home_team_metrics = [match_data['Home_Team_Score'].values[0],match_data['Home_Team_Possession'].values[0],match_data['Home_Team_Shots'].values[0],match_data['Home_Team_ShotsonTarget'].values[0],match_data['Home_Team_Passes'].values[0],match_data['Home_Team_Fouls'].values[0],match_data['Home_Team_Yellowcard'].values[0],match_data['Home_Team_Redcard'].values[0],match_data['Home_Team_Offside'].values[0],match_data['Home_Team_Corner'].values[0]]
        away_team_metrics = [match_data['Away_Team_Score'].values[0],match_data['Away_Team_Possession'].values[0],match_data['Away_Team_Shots'].values[0],match_data['Away_Team_ShotsonTarget'].values[0],match_data['Away_Team_Passes'].values[0],match_data['Away_Team_Fouls'].values[0],match_data['Away_Team_Yellowcard'].values[0],match_data['Away_Team_Redcard'].values[0],match_data['Away_Team_Offside'].values[0],match_data['Away_Team_Corner'].values[0]]
                

        table_data = {
                    "Metric" : table_metrics,
                    f"{home_input}" : home_team_metrics,
                    f"{away_input}" : away_team_metrics
                }
        
        df = pd.DataFrame(table_data)
        
        print(df.to_string(index=False))
        # print(df)

        
        
        
        #plot-1
        plt.figure(figsize=(8, 5))
        metrics_to_plot = ['Home_Team_Possession','Away_Team_Possession','Home_Team_Passes','Away_Team_Passes','Home_Team_Fouls','Away_Team_Fouls','Home_Team_Offside','Away_Team_Offside','Home_Team_Corner','Away_Team_Corner']
        metrics_names = ['Home Team Possession','Away Team Possession','Home Team Passes', 'Away Team Passes', 'Home Team Fouls', 'Away Team Fouls', 'Home Team Offsides', 'Away Team Offsides','Home Team Corners','Away Team Corners']
        metrics_data = match_data[metrics_to_plot].values
        num_metrics = len(metrics_to_plot)
        num_matches = len(match_data)
        bar_width = 0.15
        bar_space = 0.1
        
        home_color = 'blue'
        away_color = 'red'
        
        all_x = []
        
        for i in range(num_metrics):
            x = [j + (bar_width + bar_space) * i for j in range(num_matches)]
            all_x.extend(x)
            if 'Home_' in metrics_to_plot[i]:  # Check if metric is for Home team
                plt.bar(x, metrics_data[:, i], width=bar_width, label=metrics_names[i], color=home_color)
            else:
                plt.bar(x, metrics_data[:, i], width=bar_width, label=metrics_names[i], color=away_color)
        
        # Set individual labels for each bar
        plt.xticks(all_x, [name for name in metrics_names for _ in range(num_matches)], rotation=45)
        
        plt.title(f"{home_input} vs. {away_input} Metrics")
        plt.xlabel("Metrics")
        plt.ylabel("Values")
        plt.legend()
        plt.grid(True)
        plt.tight_layout()  # Adjust layout for better visualization
        plt.show()
        

        fig, ax = plt.subplots(figsize=(8, 1))
        ax.set_facecolor('black')
        ax.barh(['Possession'], match_data['Home_Team_Possession'], color='red', edgecolor='black', height=0.6)
        ax.barh(['Possession'], match_data['Away_Team_Possession'], left=match_data['Home_Team_Possession'], color='blue', edgecolor='black', height=0.6)
        ax.set_xlim(0,100)
        ax.axis('off')
        # ax.set_yticks([])
        # ax.set_ylabel('')
        ax.text(2, 0, f"{home_input}  {match_data['Home_Team_Possession'].values[0]}%", va='center', ha='left', color='white', fontweight='bold', fontsize=14)
        ax.text(98, 0, f"{match_data['Away_Team_Possession'].values[0]}%  {away_input}", va='center', ha='right', color='white', fontweight='bold', fontsize=14)
        plt.title('Possession Comparison')
        # Show the plot
        plt.tight_layout()
        plt.show()

        fig, ax = plt.subplots(figsize=(8, 1))
        ax.set_facecolor('black')
        ax.barh(['Passes'], match_data['Home_Team_Passes'], color='red', edgecolor='black', height=0.6)
        ax.barh(['Passes'], match_data['Away_Team_Passes'], left=match_data['Home_Team_Passes'], color='blue', edgecolor='black', height=0.6)
        upper_limit = match_data['Home_Team_Passes'].values[0]+match_data['Away_Team_Passes'].values[0]
        ax.set_xlim(0,upper_limit)
        ax.axis('off')
        ax.text(2, 0, f"{home_input}  {match_data['Home_Team_Passes'].values[0]}", va='center', ha='left', color='white', fontweight='bold', fontsize=14)
        ax.text(upper_limit-2, 0, f"{match_data['Away_Team_Passes'].values[0]}  {away_input}", va='center', ha='right', color='white', fontweight='bold', fontsize=14)
        plt.title('Passes Comparison')
        # Show the plot
        plt.tight_layout()
        plt.show()

        fig, ax = plt.subplots(figsize=(8, 1))
        ax.set_facecolor('black')
        ax.barh(['Fouls'], match_data['Home_Team_Fouls'], color='red', edgecolor='black', height=0.6)
        ax.barh(['Fouls'], match_data['Away_Team_Fouls'], left=match_data['Home_Team_Fouls'], color='blue', edgecolor='black', height=0.6)
        upper_limit = match_data['Home_Team_Fouls'].values[0]+match_data['Away_Team_Fouls'].values[0]
        ax.set_xlim(0,upper_limit)
        ax.axis('off')
        ax.text(2, 0, f"{home_input}  {match_data['Home_Team_Fouls'].values[0]}", va='center', ha='left', color='white', fontweight='bold', fontsize=14)
        ax.text(upper_limit-2, 0, f"{match_data['Away_Team_Fouls'].values[0]}  {away_input}", va='center', ha='right', color='white', fontweight='bold', fontsize=14)
        plt.title('Fouls Comparison')
        # Show the plot
        plt.tight_layout()
        plt.show()

        
        
    else:
        print("Invalid team name. Please enter a valid team.")
```
