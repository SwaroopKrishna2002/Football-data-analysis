```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = pd.read_csv("events_bundesliga_2022-23.csv")
```

```python
data.head()
```

```python
goal_events = data[data['event_type']=='goal']
goal_events
```

```python
goal_events.value_counts()
```

```python
goal_events = data[(data['event_type'] == 'goal') & (data['action_player_2'] != 'Disallowed goal')]
```

```python
goals_by_minute = goal_events['event_time'].value_counts().sort_index()
```

```python
plt.figure(figsize=(15, 10))
plt.bar(goals_by_minute.index, goals_by_minute.values, color='blue')
plt.xlabel('Time of Goals (Minutes)')
plt.ylabel('Number of Goals')
plt.title('Goal Bar Plot')
plt.xticks(rotation=0)
plt.show()
```
