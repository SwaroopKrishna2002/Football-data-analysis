```python
metric = ['match_id','event_type']

while True:
    player_input = input("Enter a player name (or exit to quit) : ").strip()
    
    if player_input.lower()=="exit":
        break
    
    if player_input.lower() in player_names.str.lower().values:
        player_events = data[(data['action_player_1'].str.lower()==player_input.lower()) | (data['action_player_2'].str.lower()==player_input.lower())]
        events = player_events[metric]
        print(events)
        
    else:
        print("Invalid player name. Please enter a valid player name.")
```
