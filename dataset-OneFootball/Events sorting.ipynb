```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = pd.read_csv("events_champions_league_2022-23.csv")
```

```python
data
```

```python
team_events = data.groupby(['team','event_type']).size().unstack(fill_value=0)
team_events
```

```python
plt.figure(figsize=(15,10))
team_events.plot(kind='bar',stacked=True,ax=plt.gca())

plt.xlabel("Teams")
plt.ylabel("Team Events")
plt.title("Team Behaviour")
plt.legend(title="Event Type")
plt.xticks(rotation=90)
plt.show()
```
